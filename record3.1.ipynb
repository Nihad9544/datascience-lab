```python
import matplotlib.pyplot as plt
import numpy as np

plt.subplot(2,3,1)
x=np.array([0,20])
y=np.array([0,50])
plt.xlabel("X-AXIS",color='blue')
plt.ylabel("Y-AXIS",color='blue')
plt.plot(x,y,"o")



plt.subplot(2,3,2)
x=np.array([1,20,1,20,1])
y=np.array([15,1,15,1,15])
plt.plot(x, marker = '*',color='orange')
plt.plot(y,linestyle='dashed',color='blue')

plt.subplot(2,3,3)
y=np.array([1,30,10,50])
plt.plot(y,color='blue',marker='o',ms='20',mec='r')

plt.subplot(2,3,4)
x=np.array([0,20])
y=np.array([0,50])
plt.plot(y,y)
plt.grid()

plt.subplot(2,3,5)
x=np.array(['A','B','C','D'])
y=np.array([10,30,50,40])
plt.bar(x,y,color='red')

plt.subplot(2,3,6)
x=np.array([15,30,6,9,11,8])
y=np.array([5,13,28,23,21,19])
sizes = np.array([60,70,100,200,500,1000])
colors=np.array(['red','blue','orange','pink','green','yellow'])
plt.scatter(x,y,c=colors,s=sizes,alpha=0.5)
plt.show()
```

```python
y=np.array([35,20,20,25])
mylabels=["35","20","20","25"]
myexplode=[0.2,0,0,0]
plt.pie(y,labels=mylabels,startangle=120,explode=myexplode,shadow=True)
plt.legend()
plt.show()
```
